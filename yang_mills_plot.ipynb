```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipj

# Параметры
m = 0.5                  # m = k² = 1/2
df = pd.read_csv('trajectory_2.txt', sep='\s+', comment='#', header=None, names=['t', 'x', 'y', 'px', 'py', 'energy'])

sn, cn, dn, ph = ellipj(df['t'], m)

plt.figure(figsize=(10, 5.5))
plt.plot(df['t'], cn, 'b-', lw=1.8, label=f'cn(t | {m})')
plt.plot(df['t'], df['y'], label='y(t)', color='g')
plt.plot(df['t'], df['x'], label='x(t)', color='m')

plt.title('Эллиптический косинус  cn(t | 1/2)   и сопутствующие функции')
plt.xlabel('t')
plt.ylabel('cn(t | m)')
plt.grid(True, alpha=0.35)
plt.axhline(0, color='k', lw=0.6, alpha=0.4)
plt.ylim(-1.05, 1.05)
plt.xlim(0, 3.7081493546027438)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5.5))
plt.plot(df['t'], np.abs(cn - df['y']), 'b-', lw=1.8, label=f'|cn(t | {m}) - y(t)|')
#plt.plot(df['t'], df['t'], 'b-', lw=1.8, label=f'cn(t | {m})')
plt.grid(True, alpha=0.35)
plt.axhline(0, color='k', lw=0.6, alpha=0.4)
plt.legend(loc='upper right')
#plt.xlim(0, 3.7081493546027438)
plt.tight_layout()
plt.show()
```
