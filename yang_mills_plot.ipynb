```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x_values = [
    3.14640123,
    3.06828126,
    2.81248487,
    2.30142639,
    1.17549820,
    0.78132880,
    0.75737863,
    1.13033993,
    2.28534102,
    2.80417409,
    3.06457498
]

df = pd.read_csv('trajectory_1.txt', sep='\s+', comment='#', header=None, names=['t', 'x', 'y', 'px', 'py', 'energy'])

x = df['x']
y = df['y']
px = df['px']
x = np.linspace(-3.5,3.5,1000)

plt.figure(figsize=(8, 8))
plt.plot()
plt.plot(df['x'], df['y'], marker='.', linestyle='-', color='b', markersize=1)
plt.plot(x, 1/x, marker='.', linestyle='-', color='black', markersize=1)
plt.plot(x, -1/x, marker='.', linestyle='-', color='black', markersize=1)
plt.plot(x, np.zeros(x.shape), marker='.', linestyle='-', color='black', markersize=1)
#plt.plot(df['x'], np.zeros(df['x'].shape))
#plt.scatter(x_values, np.zeros(len(x_values)), color='red', s=20, marker='o', label='пересечения y=0', zorder=10)
plt.xlim(-3.5,3.5)
plt.ylim(-3.5,3.5)
plt.title('Траектория: y vs x')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df['x'], df['px'], marker='.', linestyle='-', color='r', markersize=1)
plt.title('Фазовый портрет: px vs x')
plt.xlabel('x')
plt.ylabel('px')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df['t'], df['y'], label='y(t)', color='g')
plt.plot(df['t'], df['x'], label='x(t)', color='m')
plt.title('Зависимость координат от времени')
plt.xlabel('t')
plt.ylabel('x, y')
plt.legend()
plt.grid(True)
plt.show()
```
